```python
from datetime import date, timedelta

# Helper function to generate ordinal day suffixes (1st, 2nd, 3rd, 23rd, etc.)
def get_ordinal(day):
    if 11 <= (day % 100) <= 13:
        return f"{day}th"
    suffixes = {1: 'st', 2: 'nd', 3: 'rd'}
    return f"{day}{suffixes.get(day % 10, 'th')}"

start_date = date(2023, 2, 7)
end_date = date.today()

current_date = start_date
messages = []
current_month_year = None

while current_date <= end_date:
    # Separate each month with a heading
    month_year_heading = current_date.strftime("%B %Y").upper()
    if month_year_heading != current_month_year:
        current_month_year = month_year_heading
        if messages:
            messages.append("")
        messages.append(f"--- {current_month_year} ---")

    # Format components
    day_str = get_ordinal(current_date.day)
    month_str = current_date.strftime("%B").lower()  # e.g., "july"
    year_str = current_date.year

    # Custom line format
    messages.append(f"I loved you on -> {day_str} {month_str} {year_str} then,")
    
    current_date += timedelta(days=1)

output_text = "\n".join(messages)

# Save to a text file
filename = "i_loved_you_formatted.txt"
with open(filename, "w", encoding="utf-8") as f:
    f.write(output_text)

print(f"Done! Saved to '{filename}'.")
```
